```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding= "latin1")
df
```

1. LIMPIEZA DE DATOS

```python
df.columns
```

```python
df = df.drop(columns=["Product Description", "Product Image"])
```

```python
df.info()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df["Order Zipcode"] = df["Order Zipcode"].fillna("unkown")
```

```python
df["Order Zipcode"].isna().sum()
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
df_limpio = df.drop(columns=["Customer Email", "Customer Password","Latitude", "Longitude"])
```

```python
df_limpio["Type"] = df["Type"].str.lower()
df_limpio
```

2. IDENTIFICAR LA MAYOR TASA DE RETRASOS POR REGION

```python
late_delivery = df_limpio.groupby("Customer State")["Late_delivery_risk"].mean().sort_values(ascending= False)
late_delivery
```

3. AUDITORIA DE METODOS DE ENVIO

```python
df_limpio["shipping variance"] = df_limpio["Days for shipment (scheduled)"] - df_limpio["Days for shipping (real)"]
```

```python
efectividad_metodos_envio = df_limpio.groupby("Shipping Mode")["shipping variance"].mean().sort_values(ascending= False)
efectividad_metodos_envio
```

```python
transacciones_fraudulentas = df_limpio[df_limpio["Order Status"] == "SUSPECTED_FRAUD"]
metodos_fraude = transacciones_fraudulentas["Type"].value_counts()
metodos_fraude
```
