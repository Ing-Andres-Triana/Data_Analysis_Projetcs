```python
data.drop(columns = "Unnamed: 0", inplace=True) # Drop the  column "Unnamed: 0"
```

```python
data.columns
```

```python
data.rename(columns = {"Deatils": "Details"}, inplace=True) # Rename the column "Deatils" to "Details"
```

```python
data.columns
```

```python
import matplotlib.pyplot as plt
```

```python
data.duplicated().value_counts().plot(kind="bar") # Check for duplicate rows in the dataset   
plt.title("Duplicate Rows in the Dataset") # Set the title of the plot
plt.xlabel("Duplicate Rows") # Set the x-axis label
plt.ylabel("Count") # Set the y-axis label
plt.show()
```

```python
data.drop_duplicates(inplace=True) # Drop duplicate rows from the dataset
```

```python
data.duplicated().sum() # Check for duplicate rows in the dataset
```

```python
data = data.reset_index().drop(columns = "index")
```

```python
data.isna().sum() # Check for missing values in the dataset
```

```python
data["BrandName"] = data["BrandName"].apply(lambda x: x.capitalize())
data
```

```python
data["Details"] = data["Details"].apply(lambda x: x.capitalize())
data
```

```python
# WE FOUND Nan Values but theyre Strings
type(data["Sizes"][27927])
```

```python
data["Sizes"] = data["Sizes"].replace("Nan", "NA VALUE")
data
```

```python
data = data[data["Sizes"]!= "NA VALUE"] # Display rows where the "Sizes" column is not equal to "NA VALUE"
```

```python
data["Discount"] = data["Discount"].replace("Nan", "No discount") # Replace "Nan" values in the "Discount" column with "No discount"
```

```python
data
```

```python
data = data.reset_index().drop(columns ="index")
data
```

```python
data.to_excel("Cleaned_data.xlsx")
```

```python
data.to_csv("Cleaned_CSV.csv", index=False) # Save the cleaned dataset to a CSV file without the index column
```
