```python
data.drop (columns = ['Unnamed: 0', 'asin','product_url', 'product_photo',
       'product_num_offers', 'product_minimum_offer_price','is_amazon_choice',
         'is_prime', 'climate_pledge_friendly','sales_volume', 'delivery', 'has_variations',], inplace = True) # usamos la funcion drop para eliminar columnas del dataframe
```

Vamos a renombrar las columnas poniendoles mayusculas a la primera letra

```python

data_new_column_names = []

for x in data.columns:
    data_new_column_names.append(x.capitalize())
```

```python
data.columns =data_new_column_names #asiganamos los nuevos nombres de columnas al dataframe
```

```python
data.head(2)#verificamos los cambios
```

VAMOS A ELIMINAR NA VALUES

```python
data.isna().sum() # usamos la funcion isna para ver si hay valores nulos en el dataframe
```

```python
data.shape # usamos la funcion shape para ver el tamaño del dataframe
```

```python
#como hay muchos valores nulos para tan pocos registros vamos a tomar solo las columnas que nos sirve

data = data[["Product_title", "Product_price", "Currency", "Product_star_rating", "Product_num_ratings", "Is_best_seller"]]
data
```

```python
data.isna().sum() # usamos la funcion isna para ver si hay valores nulos en el dataframe
```

```python
data.dropna(inplace = True) # usamos la funcion dropna para eliminar filas con valores nulos del dataframe
```

```python
data.isna().sum() # usamos la funcion isna para verificar si hay valores nulos en el dataframe
```

VAMOS A VERIFICAR SI PODEMOS HACER MAS CAMBIOS

```python
data.head(3)
```

vAMOS A TRANSFORMAR LOS VALORES DE LA COLUMNA Product_price A TIPO FLOAT PARA PODER HACER ANALISIS DE DATOS

```python
data["Product_price"]= data["Product_price"].str.replace("$", "").astype(float) # usamos la funcion str.replace para eliminar el simbolo de dolar y luego usamos astype para convertir la columna a tipo float
```

```python
data.head(2)
```

VAMOS A TRANSFORMAR LOS DATOS DE LA COLUMNA is_best_seller asignando valores como 0 y 1


```python
data["Is_best_seller"]= data["Is_best_seller"].apply(lambda x: 1 if x == True else 0) # usamos la funcion apply para aplicar una funcion lambda a la columna Is_best_seller y convertir los valores True y False a 1 y 0 respectivamente
```
